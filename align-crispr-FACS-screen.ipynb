# CRISPR-screen alignment
This notebook builds a Bowtie2 index from a gRNA library, aligns single-end FASTQ files with exact-match settings, converts, sorts, and indexes BAM files, generates per-guide counts, and combines counts into a table.


```python
import os
from pathlib import Path

PROJECT_DIR = Path("/path/to/project")
BOWTIE2_MODULE = "Bowtie2/2.3.5.1"
SAMTOOLS_MODULE = "samtools"

os.environ.update({
    "PROJECT_DIR": str(PROJECT_DIR),
    "BOWTIE2_MODULE": BOWTIE2_MODULE,
    "SAMTOOLS_MODULE": SAMTOOLS_MODULE,
})
```

### Set permissions


```python
%%bash
chmod -R 774 "$PROJECT_DIR"
```

### Start interactive node


```python
%%bash
srun --pty --mem=64G bash -i
```

### Build alignment library


```python
%%bash
cd "$PROJECT_DIR"
module load "$BOWTIE2_MODULE"
bowtie2-build -f library-fasta-file.fa gRNA_library
```

### Align reads


```python
%%bash
cd "$PROJECT_DIR"
module load "$BOWTIE2_MODULE"

bowtie2 --end-to-end --very-sensitive -L 21 -N 0 --no-1upfront --trim-to 3:21 -x gRNA_library -U 1H.fastq.gz -S 1H_alignment.sam --no-unal --score-min L,0,-0.01 ;
bowtie2 --end-to-end --very-sensitive -L 21 -N 0 --no-1upfront --trim-to 3:21 -x gRNA_library -U 1L.fastq.gz -S 1L_alignment.sam --no-unal --score-min L,0,-0.01 ;
bowtie2 --end-to-end --very-sensitive -L 21 -N 0 --no-1upfront --trim-to 3:21 -x gRNA_library -U 1U.fastq.gz -S 1U_alignment.sam --no-unal --score-min L,0,-0.01 ;
bowtie2 --end-to-end --very-sensitive -L 21 -N 0 --no-1upfront --trim-to 3:21 -x gRNA_library -U 2H.fastq.gz -S 2H_alignment.sam --no-unal --score-min L,0,-0.01 ;
bowtie2 --end-to-end --very-sensitive -L 21 -N 0 --no-1upfront --trim-to 3:21 -x gRNA_library -U 2L.fastq.gz -S 2L_alignment.sam --no-unal --score-min L,0,-0.01 ;
bowtie2 --end-to-end --very-sensitive -L 21 -N 0 --no-1upfront --trim-to 3:21 -x gRNA_library -U 2U.fastq.gz -S 2U_alignment.sam --no-unal --score-min L,0,-0.01 ;
bowtie2 --end-to-end --very-sensitive -L 21 -N 0 --no-1upfront --trim-to 3:21 -x gRNA_library -U 3H.fastq.gz -S 3H_alignment.sam --no-unal --score-min L,0,-0.01 ;
bowtie2 --end-to-end --very-sensitive -L 21 -N 0 --no-1upfront --trim-to 3:21 -x gRNA_library -U 3L.fastq.gz -S 3L_alignment.sam --no-unal --score-min L,0,-0.01 ;
bowtie2 --end-to-end --very-sensitive -L 21 -N 0 --no-1upfront --trim-to 3:21 -x gRNA_library -U 3U.fastq.gz -S 3U_alignment.sam --no-unal --score-min L,0,-0.01 ;
bowtie2 --end-to-end --very-sensitive -L 21 -N 0 --no-1upfront --trim-to 3:21 -x gRNA_library -U 4H.fastq.gz -S 4H_alignment.sam --no-unal --score-min L,0,-0.01 ;
bowtie2 --end-to-end --very-sensitive -L 21 -N 0 --no-1upfront --trim-to 3:21 -x gRNA_library -U 4L.fastq.gz -S 4L_alignment.sam --no-unal --score-min L,0,-0.01 ;
bowtie2 --end-to-end --very-sensitive -L 21 -N 0 --no-1upfront --trim-to 3:21 -x gRNA_library -U 4U.fastq.gz -S 4U_alignment.sam --no-unal --score-min L,0,-0.01
```

### Convert, sort, and index alignments


```python
%%bash
cd "$PROJECT_DIR"
module load "$SAMTOOLS_MODULE"

samtools view -bS 1H_alignment.sam > 1H_alignment.bam ;
samtools view -bS 1L_alignment.sam > 1L_alignment.bam ;
samtools view -bS 1U_alignment.sam > 1U_alignment.bam ;
samtools view -bS 2H_alignment.sam > 2H_alignment.bam ;
samtools view -bS 2L_alignment.sam > 2L_alignment.bam ;
samtools view -bS 2U_alignment.sam > 2U_alignment.bam ;
samtools view -bS 3H_alignment.sam > 3H_alignment.bam ;
samtools view -bS 3L_alignment.sam > 3L_alignment.bam ;
samtools view -bS 3U_alignment.sam > 3U_alignment.bam ;
samtools view -bS 4H_alignment.sam > 4H_alignment.bam ;
samtools view -bS 4L_alignment.sam > 4L_alignment.bam ;
samtools view -bS 4U_alignment.sam > 4U_alignment.bam ;

samtools sort -@ 32 -m 2G 1H_alignment.bam -o 1H_alignment.sorted.bam ;
samtools sort -@ 32 -m 2G 1L_alignment.bam -o 1L_alignment.sorted.bam ;
samtools sort -@ 32 -m 2G 1U_alignment.bam -o 1U_alignment.sorted.bam ;
samtools sort -@ 32 -m 2G 2H_alignment.bam -o 2H_alignment.sorted.bam ;
samtools sort -@ 32 -m 2G 2L_alignment.bam -o 2L_alignment.sorted.bam ;
samtools sort -@ 32 -m 2G 2U_alignment.bam -o 2U_alignment.sorted.bam ;
samtools sort -@ 32 -m 2G 3H_alignment.bam -o 3H_alignment.sorted.bam ;
samtools sort -@ 32 -m 2G 3L_alignment.bam -o 3L_alignment.sorted.bam ;
samtools sort -@ 32 -m 2G 3U_alignment.bam -o 3U_alignment.sorted.bam ;
samtools sort -@ 32 -m 2G 4H_alignment.bam -o 4H_alignment.sorted.bam ;
samtools sort -@ 32 -m 2G 4L_alignment.bam -o 4L_alignment.sorted.bam ;
samtools sort -@ 32 -m 2G 4U_alignment.bam -o 4U_alignment.sorted.bam ;

samtools index 1H_alignment.sorted.bam ;
samtools index 1L_alignment.sorted.bam ;
samtools index 1U_alignment.sorted.bam ;
samtools index 2H_alignment.sorted.bam ;
samtools index 2L_alignment.sorted.bam ;
samtools index 2U_alignment.sorted.bam ;
samtools index 3H_alignment.sorted.bam ;
samtools index 3L_alignment.sorted.bam ;
samtools index 3U_alignment.sorted.bam ;
samtools index 4H_alignment.sorted.bam ;
samtools index 4L_alignment.sorted.bam ;
samtools index 4U_alignment.sorted.bam
```

### Generate and combine counts


```python
%%bash
cd "$PROJECT_DIR"
module load "$SAMTOOLS_MODULE"

samtools idxstats 1H_alignment.sorted.bam | awk '{print $1"\t"$3}' | grep -v "^\*" > 1H.counts ;
samtools idxstats 1L_alignment.sorted.bam | awk '{print $1"\t"$3}' | grep -v "^\*" > 1L.counts ;
samtools idxstats 1U_alignment.sorted.bam | awk '{print $1"\t"$3}' | grep -v "^\*" > 1U.counts ;
samtools idxstats 2H_alignment.sorted.bam | awk '{print $1"\t"$3}' | grep -v "^\*" > 2H.counts ;
samtools idxstats 2L_alignment.sorted.bam | awk '{print $1"\t"$3}' | grep -v "^\*" > 2L.counts ;
samtools idxstats 2U_alignment.sorted.bam | awk '{print $1"\t"$3}' | grep -v "^\*" > 2U.counts ;
samtools idxstats 3H_alignment.sorted.bam | awk '{print $1"\t"$3}' | grep -v "^\*" > 3H.counts ;
samtools idxstats 3L_alignment.sorted.bam | awk '{print $1"\t"$3}' | grep -v "^\*" > 3L.counts ;
samtools idxstats 3U_alignment.sorted.bam | awk '{print $1"\t"$3}' | grep -v "^\*" > 3U.counts ;
samtools idxstats 4H_alignment.sorted.bam | awk '{print $1"\t"$3}' | grep -v "^\*" > 4H.counts ;
samtools idxstats 4L_alignment.sorted.bam | awk '{print $1"\t"$3}' | grep -v "^\*" > 4L.counts ;
samtools idxstats 4U_alignment.sorted.bam | awk '{print $1"\t"$3}' | grep -v "^\*" > 4U.counts

paste 1H.counts 1L.counts 1U.counts \
2H.counts 2L.counts 2U.counts \
3H.counts 3L.counts 3U.counts \
4H.counts 4L.counts 4U.counts > countstable.txt
```
