# CRISPR-screen alignment
This notebook builds a Bowtie2 index from a gRNA library, aligns single-end FASTQ files with exact-match settings, converts, sorts, and indexes BAM files, generates per-guide counts, and combines counts into a table.


```python
import os
from pathlib import Path

PROJECT_DIR = Path("/path/to/project")
BOWTIE2_MODULE = "Bowtie2/2.3.5.1"
SAMTOOLS_MODULE = "samtools"

os.environ.update({
    "PROJECT_DIR": str(PROJECT_DIR),
    "BOWTIE2_MODULE": BOWTIE2_MODULE,
    "SAMTOOLS_MODULE": SAMTOOLS_MODULE,
})
```

## Set permissions


```python
%%bash
chmod -R 774 "$PROJECT_DIR"
```

## Start interactive node


```python
%%bash
srun --pty --mem=64G bash -i
```

## Build alignment library


```python
%%bash
cd "$PROJECT_DIR"
module load "$BOWTIE2_MODULE"
bowtie2-build -f mmlib_21bp_all.fa gRNA_library
```

## Align reads


```python
%%bash
cd "$PROJECT_DIR"
module load "$BOWTIE2_MODULE"

bowtie2 --end-to-end --very-sensitive -L 21 -N 0 --no-1mm-upfront --trim-to 3:21 -x gRNA_library -U mm-1H.fastq.gz -S mm-1H_alignment.sam --no-unal --score-min L,0,-0.01 ;
bowtie2 --end-to-end --very-sensitive -L 21 -N 0 --no-1mm-upfront --trim-to 3:21 -x gRNA_library -U mm-1L.fastq.gz -S mm-1L_alignment.sam --no-unal --score-min L,0,-0.01 ;
bowtie2 --end-to-end --very-sensitive -L 21 -N 0 --no-1mm-upfront --trim-to 3:21 -x gRNA_library -U mm-1U.fastq.gz -S mm-1U_alignment.sam --no-unal --score-min L,0,-0.01 ;
bowtie2 --end-to-end --very-sensitive -L 21 -N 0 --no-1mm-upfront --trim-to 3:21 -x gRNA_library -U mm-2H.fastq.gz -S mm-2H_alignment.sam --no-unal --score-min L,0,-0.01 ;
bowtie2 --end-to-end --very-sensitive -L 21 -N 0 --no-1mm-upfront --trim-to 3:21 -x gRNA_library -U mm-2L.fastq.gz -S mm-2L_alignment.sam --no-unal --score-min L,0,-0.01 ;
bowtie2 --end-to-end --very-sensitive -L 21 -N 0 --no-1mm-upfront --trim-to 3:21 -x gRNA_library -U mm-2U.fastq.gz -S mm-2U_alignment.sam --no-unal --score-min L,0,-0.01 ;
bowtie2 --end-to-end --very-sensitive -L 21 -N 0 --no-1mm-upfront --trim-to 3:21 -x gRNA_library -U mm-3H.fastq.gz -S mm-3H_alignment.sam --no-unal --score-min L,0,-0.01 ;
bowtie2 --end-to-end --very-sensitive -L 21 -N 0 --no-1mm-upfront --trim-to 3:21 -x gRNA_library -U mm-3L.fastq.gz -S mm-3L_alignment.sam --no-unal --score-min L,0,-0.01 ;
bowtie2 --end-to-end --very-sensitive -L 21 -N 0 --no-1mm-upfront --trim-to 3:21 -x gRNA_library -U mm-3U.fastq.gz -S mm-3U_alignment.sam --no-unal --score-min L,0,-0.01 ;
bowtie2 --end-to-end --very-sensitive -L 21 -N 0 --no-1mm-upfront --trim-to 3:21 -x gRNA_library -U mm-4H.fastq.gz -S mm-4H_alignment.sam --no-unal --score-min L,0,-0.01 ;
bowtie2 --end-to-end --very-sensitive -L 21 -N 0 --no-1mm-upfront --trim-to 3:21 -x gRNA_library -U mm-4L.fastq.gz -S mm-4L_alignment.sam --no-unal --score-min L,0,-0.01 ;
bowtie2 --end-to-end --very-sensitive -L 21 -N 0 --no-1mm-upfront --trim-to 3:21 -x gRNA_library -U mm-4U.fastq.gz -S mm-4U_alignment.sam --no-unal --score-min L,0,-0.01
```

## Convert, sort, and index alignments


```python
%%bash
cd "$PROJECT_DIR"
module load "$SAMTOOLS_MODULE"

samtools view -bS mm-1H_alignment.sam > mm-1H_alignment.bam ;
samtools view -bS mm-1L_alignment.sam > mm-1L_alignment.bam ;
samtools view -bS mm-1U_alignment.sam > mm-1U_alignment.bam ;
samtools view -bS mm-2H_alignment.sam > mm-2H_alignment.bam ;
samtools view -bS mm-2L_alignment.sam > mm-2L_alignment.bam ;
samtools view -bS mm-2U_alignment.sam > mm-2U_alignment.bam ;
samtools view -bS mm-3H_alignment.sam > mm-3H_alignment.bam ;
samtools view -bS mm-3L_alignment.sam > mm-3L_alignment.bam ;
samtools view -bS mm-3U_alignment.sam > mm-3U_alignment.bam ;
samtools view -bS mm-4H_alignment.sam > mm-4H_alignment.bam ;
samtools view -bS mm-4L_alignment.sam > mm-4L_alignment.bam ;
samtools view -bS mm-4U_alignment.sam > mm-4U_alignment.bam ;

samtools sort -@ 32 -m 2G mm-1H_alignment.bam -o mm-1H_alignment.sorted.bam ;
samtools sort -@ 32 -m 2G mm-1L_alignment.bam -o mm-1L_alignment.sorted.bam ;
samtools sort -@ 32 -m 2G mm-1U_alignment.bam -o mm-1U_alignment.sorted.bam ;
samtools sort -@ 32 -m 2G mm-2H_alignment.bam -o mm-2H_alignment.sorted.bam ;
samtools sort -@ 32 -m 2G mm-2L_alignment.bam -o mm-2L_alignment.sorted.bam ;
samtools sort -@ 32 -m 2G mm-2U_alignment.bam -o mm-2U_alignment.sorted.bam ;
samtools sort -@ 32 -m 2G mm-3H_alignment.bam -o mm-3H_alignment.sorted.bam ;
samtools sort -@ 32 -m 2G mm-3L_alignment.bam -o mm-3L_alignment.sorted.bam ;
samtools sort -@ 32 -m 2G mm-3U_alignment.bam -o mm-3U_alignment.sorted.bam ;
samtools sort -@ 32 -m 2G mm-4H_alignment.bam -o mm-4H_alignment.sorted.bam ;
samtools sort -@ 32 -m 2G mm-4L_alignment.bam -o mm-4L_alignment.sorted.bam ;
samtools sort -@ 32 -m 2G mm-4U_alignment.bam -o mm-4U_alignment.sorted.bam ;

samtools index mm-1H_alignment.sorted.bam ;
samtools index mm-1L_alignment.sorted.bam ;
samtools index mm-1U_alignment.sorted.bam ;
samtools index mm-2H_alignment.sorted.bam ;
samtools index mm-2L_alignment.sorted.bam ;
samtools index mm-2U_alignment.sorted.bam ;
samtools index mm-3H_alignment.sorted.bam ;
samtools index mm-3L_alignment.sorted.bam ;
samtools index mm-3U_alignment.sorted.bam ;
samtools index mm-4H_alignment.sorted.bam ;
samtools index mm-4L_alignment.sorted.bam ;
samtools index mm-4U_alignment.sorted.bam
```

## Generate and combine counts


```python
%%bash
cd "$PROJECT_DIR"
module load "$SAMTOOLS_MODULE"

samtools idxstats mm-1H_alignment.sorted.bam | awk '{print $1"\t"$3}' | grep -v "^\*" > mm-1H.counts ;
samtools idxstats mm-1L_alignment.sorted.bam | awk '{print $1"\t"$3}' | grep -v "^\*" > mm-1L.counts ;
samtools idxstats mm-1U_alignment.sorted.bam | awk '{print $1"\t"$3}' | grep -v "^\*" > mm-1U.counts ;
samtools idxstats mm-2H_alignment.sorted.bam | awk '{print $1"\t"$3}' | grep -v "^\*" > mm-2H.counts ;
samtools idxstats mm-2L_alignment.sorted.bam | awk '{print $1"\t"$3}' | grep -v "^\*" > mm-2L.counts ;
samtools idxstats mm-2U_alignment.sorted.bam | awk '{print $1"\t"$3}' | grep -v "^\*" > mm-2U.counts ;
samtools idxstats mm-3H_alignment.sorted.bam | awk '{print $1"\t"$3}' | grep -v "^\*" > mm-3H.counts ;
samtools idxstats mm-3L_alignment.sorted.bam | awk '{print $1"\t"$3}' | grep -v "^\*" > mm-3L.counts ;
samtools idxstats mm-3U_alignment.sorted.bam | awk '{print $1"\t"$3}' | grep -v "^\*" > mm-3U.counts ;
samtools idxstats mm-4H_alignment.sorted.bam | awk '{print $1"\t"$3}' | grep -v "^\*" > mm-4H.counts ;
samtools idxstats mm-4L_alignment.sorted.bam | awk '{print $1"\t"$3}' | grep -v "^\*" > mm-4L.counts ;
samtools idxstats mm-4U_alignment.sorted.bam | awk '{print $1"\t"$3}' | grep -v "^\*" > mm-4U.counts

paste mm-1H.counts mm-1L.counts mm-1U.counts \
mm-2H.counts mm-2L.counts mm-2U.counts \
mm-3H.counts mm-3L.counts mm-3U.counts \
mm-4H.counts mm-4L.counts mm-4U.counts > countstable.txt
```
